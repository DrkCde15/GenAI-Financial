```python
from pyspark.sql import functions as F

# Leitura dos dados tratados da camada silver
df_silver = spark.table('db_analytics.silver_bitcoin')

# Padronização para o modelo gold: nomes finais, tipos e ordem das colunas
df_gold = df_silver.select(
    F.col('semana_inicio').cast('date').alias('semana_inicio'),
    F.col('semana_fim').cast('date').alias('semana_fim'),
    F.col('abertura').alias('abertura'),
    F.col('fechamento').alias('fechamento'),
    F.col('minimo').alias('minimo'),
    F.col('maximo').alias('maximo'),
    F.col('variação').alias('variacao_pct'),
    F.col('volume').alias('volume_medio'),
)

# Gravação na camada gold
(
    df_gold
    .write
    .format('delta')
    .mode('overwrite')
    .saveAsTable('db_analytics.gold_bitcoin')
)

print(df_gold.count())
display(spark.table('db_analytics.gold_bitcoin'))
```
