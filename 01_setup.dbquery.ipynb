```sql
CREATE DATABASE IF NOT EXISTS db_analytics;

CREATE TABLE IF NOT EXISTS db_analytics.bronze_bitcoin(
    data DATE,
    abertura DOUBLE,
    fechamento DOUBLE,
    minimo DOUBLE,
    maximo DOUBLE,
    volume DOUBLE
)
USING DELTA;

CREATE TABLE IF NOT EXISTS db_analytics.gold_bitcoin(
    semana_inicio DATE,
    semana_fim DATE,
    abertura DOUBLE,
    fechamento DOUBLE,
    minimo DOUBLE,
    maximo DOUBLE,
    variacao_pct DOUBLE,
    volume_medio DOUBLE
)
USING DELTA;


SELECT * FROM db_analytics.silver_bitcoin LIMIT 10
```
